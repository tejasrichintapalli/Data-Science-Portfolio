```python
# Drop columns
df1 = df1.drop(columns=["ReceptionistID", "CheckInTime"])

print(df1)
```

```python
# Total bill amount for each department
total_bill = df1.groupby("Department")["BillAmount"].sum()

print(total_bill)
```

```python
df1.duplicated().sum()
```

```python
# Remove duplicate PatientID records
df = df.drop_duplicates(subset="PatientID")

print(df)
```

```python
# Find the mean bill amount
mean_bill = df1["BillAmount"].mean()

# Fill missing values with the mean
df1["BillAmount"] = df1["BillAmount"].fillna(mean_bill)

print(df1)
```

```python
merged_df = pd.merge(df1, df, on="PatientID")

print(merged_df)
```

```python
new_patients = pd.DataFrame({
    "PatientID": [103, 104],
    "Name": ["Amit", "Sneha"],
    "Department": ["Orthopedic", "Cardiology"]
})

df = pd.concat([df, new_patients], ignore_index=True)

print(df)
```

```python
billing_info = pd.DataFrame({
    "InsuranceCovered": [2000, 3000, 1500],
    "FinalAmount": [3000, 4000, 2500]
})

df = pd.concat([df, billing_info], axis=1)

print(df)
```
